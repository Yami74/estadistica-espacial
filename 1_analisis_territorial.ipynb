```python
# Análisis Territorial Agropecuario (ENA - Perú)
#Este notebook procesa los microdatos de la Encuesta Nacional Agropecuaria, aplica el algoritmo de optimización de Jenks para clasificación altitudinal/productiva y utiliza K-Means para descubrir clústeres espaciales territoriales.
```

```python
import pandas as pd
import numpy as np
import geopandas as gpd
from shapely.geometry import Point
import jenkspy
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import folium

# Configuración visual
sns.set_theme(style="whitegrid")
import warnings
warnings.filterwarnings('ignore')
```

```python
print("Cargando y limpiando datos...")
# Cambia 'datos_ena.csv' por el nombre real de tu archivo
df = pd.read_csv('ENA_2014_2024.csv', low_memory=False)

# --- CORRECCIÓN DEL CRASH: Forzar conversión a números ---
columnas_numericas = ['LATITUD', 'LONGITUD', 'ALTITUD', 'P104_SUP_ha']

for col in columnas_numericas:
    if col in df.columns:
        # 'coerce' transforma los textos inválidos o espacios en blanco en NaN
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Ahora sí limpiamos los nulos con seguridad
df = df.dropna(subset=columnas_numericas)

# Filtro geográfico para Perú (aproximado) y altitud lógica
df = df[
    (df['LATITUD'] >= -19.0) & (df['LATITUD'] <= 0.0) &
    (df['LONGITUD'] >= -82.0) & (df['LONGITUD'] <= -68.0) &
    (df['ALTITUD'] >= 0) & (df['ALTITUD'] <= 6000)
]

# Crear superficie total (P104 + P105 si existe)
p105 = df['P105_SUP_ha'].fillna(0) if 'P105_SUP_ha' in df.columns else 0
# Asegurar que P105 también sea numérico si existe
if 'P105_SUP_ha' in df.columns:
    p105 = pd.to_numeric(p105, errors='coerce').fillna(0)

df['SUP_TOTAL_HA'] = df['P104_SUP_ha'] + p105

print(f"Datos limpios. Total de registros útiles: {df.shape[0]}")
```
